```python
import requests
import json

def consultar_cep(cep: str) -> dict:
    """Consulta CEP na API ViaCEP e retorna um dicionário com os dados."""
    # Remove traços e espaços
    cep = cep.replace("-", "").strip()
    
    url = f"https://viacep.com.br/ws/{cep}/json/"
    try:
        response = requests.get(url, timeout=10)
        response.raise_for_status()  # levanta erro se status != 200
    except requests.RequestException as e:
        raise SystemExit(f"Erro na requisição: {e}")
    
    data = response.json()

    if "erro" in data:
        raise ValueError("CEP não encontrado")
    
    return data

if __name__ == "__main__":
    # Input do usuário
    cep_input = input("Informe o CEP (somente números ou com traço): ")
    
    try:
        resultado = consultar_cep(cep_input)
        # Exibe JSON formatado
        print(json.dumps(resultado, indent=2, ensure_ascii=False))
    except ValueError as ve:
        print(f"Erro: {ve}")
    except SystemExit as se:
        print(se)
```
